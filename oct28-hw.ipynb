```python
import matplotlib.pyplot as plt
import pandas as pd

# Load and prepare data
trends = pd.read_csv("ai_tools_comparison.csv")
trends["week"] = pd.to_datetime(trends["week"])
trends.set_index("week", inplace=True)

# 1. Most consistent tool
weekly_growth = trends.pct_change().fillna(0) * 100
most_consistent_tool = weekly_growth.std().idxmin()
print(f"Most consistent tool: {most_consistent_tool}")

# 2. Visualize interest & pinpoint ChatGPT's major dip
trends.plot(
    figsize=(10, 5),
    title="Interest Levels of ChatGPT, Gemini, and Microsoft Copilot Over Time",
)
plt.xlabel("Week")
plt.ylabel("Interest Level")
plt.tight_layout()
plt.show()

gpt_dip = "December 2023"
print(f"ChatGPT largest dip: {gpt_dip}")

# 3. Best month across all tools ('M' for older pandas versions)
monthly_avg = trends.resample("M").mean()
best_month = monthly_avg.mean(axis=1).idxmax().strftime("%B")
print(f"Month with highest average interest: {best_month}")
```
